```python
print('iron pulse fitness club')

class Member:
    def __init__(self, name, age, plan):
        self.name = name
        self.age = age
        self.plan = plan
        self.visits = []

m1 = Member('aarav', 24, 'Monthly')
m2 = Member('diya', 28, 'Quarterly')
m3 = Member('kabir', 22, 'Annual')
m4 = Member('ananya', 31, 'Monthly')
m5 = Member('rohan', 26, 'Quarterly')

for m in [m1, m2, m3, m4, m5]:
    print(m.name, m.age, m.plan)
```

```python
def check_in(self, date_text):
    self.visits.append(date_text)

def visit_count(self):
    return len(self.visits)

def plan_price(self):
    if self.plan == 'Monthly':
        return 1200
    if self.plan == 'Quarterly':
        return 3300
    if self.plan == 'Annual':
        return 11000
    return 0

Member.check_in = check_in
Member.visit_count = visit_count
Member.plan_price = plan_price

m1.check_in('2026-03-01')
m1.check_in('2026-03-03')
m1.check_in('2026-03-05')

m2.check_in('2026-03-02')
m2.check_in('2026-03-04')

for m in [m1, m2, m3, m4, m5]:
    print(m.name, m.visit_count(), m.plan_price())
```
