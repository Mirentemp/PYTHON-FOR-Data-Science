```python
import urllib.request
import json

url = 'https://jsonplaceholder.typicode.com/users/3'
with urllib.request.urlopen(url) as response:
    raw_bytes = response.read()
    user_data = json.loads(raw_bytes.decode('utf-8'))

print(list(user_data.keys()))
print(user_data['name'])
print(user_data['email'])
print(user_data['address']['city'])
print(user_data['company']['name'])
```

```python
import urllib.parse
import json

base_url = 'https://api.bikes-example.in/v1'
headers = {
    'Content-Type': 'application/json',
    'Authorization': 'Bearer sample_token_123'
}

query = urllib.parse.urlencode({'city': 'Surat', 'max_price': 60000})

bike_api_plan = [
    {
        'action': 'list_bikes',
        'method': 'GET',
        'url': f'{base_url}/bikes?{query}',
        'headers': headers,
        'body': None
    },
    {
        'action': 'get_one_bike',
        'method': 'GET',
        'url': f'{base_url}/bikes/101',
        'headers': headers,
        'body': None
    },
    {
        'action': 'add_listing',
        'method': 'POST',
        'url': f'{base_url}/bikes',
        'headers': headers,
        'body': {'brand': 'honda', 'model': 'activa 6g', 'city': 'Surat', 'price': 55000}
    },
    {
        'action': 'change_price',
        'method': 'PUT',
        'url': f'{base_url}/bikes/101',
        'headers': headers,
        'body': {'price': 52000}
    },
    {
        'action': 'delete_listing',
        'method': 'DELETE',
        'url': f'{base_url}/bikes/101',
        'headers': headers,
        'body': None
    }
]

for call in bike_api_plan:
    print(call['method'], call['url'])

with open('bike_api_plan.json', 'w', encoding='utf-8') as f:
    json.dump(bike_api_plan, f, indent=2)
```

```python
import urllib.request
import urllib.error
import json
import time

endpoints = [
    'https://jsonplaceholder.typicode.com/posts?userId=2',
    'https://jsonplaceholder.typicode.com/posts/1',
    'https://jsonplaceholder.typicode.com/todos/5',
    'https://jsonplaceholder.typicode.com/posts/9999'
]

call_log = []

for idx, ep in enumerate(endpoints):
    if idx > 0:
        time.sleep(1)
    try:
        with urllib.request.urlopen(ep) as resp:
            status = resp.status
            content_type = resp.headers.get('Content-Type', '')
            payload = json.loads(resp.read().decode('utf-8'))
            count = len(payload) if isinstance(payload, list) else 1
            call_log.append({
                'url': ep,
                'status': status,
                'content_type': content_type,
                'item_count': count
            })
    except urllib.error.HTTPError as e:
        call_log.append({
            'url': ep,
            'status': e.code,
            'content_type': e.headers.get('Content-Type', '') if e.headers else '',
            'item_count': 0
        })

with open('api_call_log.json', 'w', encoding='utf-8') as f:
    json.dump(call_log, f, indent=2)

for entry in call_log:
    print(entry)
```
