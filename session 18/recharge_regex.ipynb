```python
import re

sms_lines = [
    'recharge of rs.299 for 9000000001 done on 14/03/2026. support: help@fastnet-example.com',
    'recharge of rs.199 for 9000000002 done on 14/03/2026. support: care@swiftpay-example.org',
    'recharge of rs.599 for 9000000003 done on 15/03/2026. support: info@bharatnet-example.in',
    'recharge of rs.99 for 9000000004 done on 15/03/2026. support: billing@skyline-example.net',
    'recharge of rs.499 for 9000000005 done on 16/03/2026. support: help@fastnet-example.com',
    'recharge of rs.19 for 9000000006 done on 16/03/2026. support: care@swiftpay-example.org',
    'recharge of rs.999 for 9000000007 done on 17/03/2026. support: info@bharatnet-example.in',
    'recharge of rs.249 for 9000000008 done on 17/03/2026. support: billing@skyline-example.net',
    'recharge of rs.399 for 9000000009 done on 18/03/2026. support: help@fastnet-example.com',
    'recharge of rs.149 for 9000000010 done on 18/03/2026. support: care@swiftpay-example.org'
]

with open('recharge_sms.txt', 'w', encoding='utf-8') as f:
    for line in sms_lines:
        f.write(line + '\n')

with open('recharge_sms.txt', 'r', encoding='utf-8') as f:
    content = f.read()

mobiles = re.findall(r'(?<!\d)[6-9]\d{9}(?!\d)', content)
emails = re.findall(r'[a-z0-9._-]+@[a-z0-9.-]+\.[a-z]{2,}', content)

print(mobiles, len(mobiles))
print(emails, len(emails))
```

```python
date_match = re.search(r'(\d{2}/\d{2}/\d{4})', content)
print(date_match.group(1))

amount_strs = re.findall(r'rs\.(\d+)', content)
amounts = [int(a) for a in amount_strs]

print(sum(amounts))
print(max(amounts))
```

```python
masked_content = re.sub(r'([6-9]\d{6})(\d{3})', r'xxxxxxx\2', content)
cleaned_content = re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\3-\2-\1', masked_content)

with open('recharge_sms_clean.txt', 'w', encoding='utf-8') as f:
    f.write(cleaned_content)

first_three = cleaned_content.splitlines()[:3]
for line in first_three:
    print(line)
```

```python
def validate_contact(text):
    if re.fullmatch(r'[a-z0-9._-]+@[a-z0-9.-]+\.[a-z]{2,}', text):
        return 'email'
    if re.fullmatch(r'[6-9]\d{9}', text):
        return 'mobile'
    return 'invalid'

samples = [
    'help@fastnet-example.com',
    'a.b@site.in',
    '9000000001',
    '5000000001',
    'name@@x.com',
    '90000',
    'info@host',
    'user_1@example.org'
]

results = []
for s in samples:
    res = validate_contact(s)
    results.append(res)
    print(res)

with open('regex_validation.txt', 'w', encoding='utf-8') as f:
    for r in results:
        f.write(r + '\n')
```
