```python
class Member:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def monthly_fee(self):
        return 1200

    def describe(self):
        return f'{self.name} (age {self.age})'

class StudentMember(Member):
    def __init__(self, name, age):
        super().__init__(name, age)

    def monthly_fee(self):
        return int(super().monthly_fee() * 0.8)

class PremiumMember(Member):
    def __init__(self, name, age, trainer_name):
        super().__init__(name, age)
        self.trainer_name = trainer_name

    def monthly_fee(self):
        return super().monthly_fee() + 800

sample_list = [Member('aarav', 28), StudentMember('diya', 20), PremiumMember('kabir', 32, 'rohan')]
for item in sample_list:
    print(type(item).__name__, item.monthly_fee())
```

```python
def total_monthly_income(members):
    return sum(mem.monthly_fee() for mem in members)

group = [
    Member('aarav', 28),
    Member('ananya', 30),
    Member('yash', 26),
    StudentMember('diya', 20),
    StudentMember('tara', 19),
    StudentMember('om', 21),
    PremiumMember('kabir', 32, 'rohan'),
    PremiumMember('meera', 29, 'vikram')
]

print(total_monthly_income(group))

student_count = sum(1 for mem in group if isinstance(mem, StudentMember))
print(student_count)

for cls in [Member, StudentMember, PremiumMember]:
    matched = [mem for mem in group if type(mem) is cls]
    cls_income = sum(mem.monthly_fee() for mem in matched)
    print(f'{cls.__name__:<15} {len(matched):<6} {cls_income}')
```
