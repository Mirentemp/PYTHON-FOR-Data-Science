```python
price_text = '48500'
accessories_text = '1250.50'
insurance_text = '3299.50'
rto_text = '1800'
km_text = '32000'
owner_count_text = '2'

print("Before conversion:")
print(price_text, type(price_text))
print(accessories_text, type(accessories_text))
print(insurance_text, type(insurance_text))
print(rto_text, type(rto_text))
print(km_text, type(km_text))
print(owner_count_text, type(owner_count_text))

price = float(price_text)
accessories = float(accessories_text)
insurance = float(insurance_text)
rto = float(rto_text)
km = int(km_text)
owner_count = int(owner_count_text)

print("\nAfter conversion:")
print(price, type(price))
print(accessories, type(accessories))
print(insurance, type(insurance))
print(rto, type(rto))
print(km, type(km))
print(owner_count, type(owner_count))

on_road_total = round(price + accessories + insurance + rto, 2)
print("\nOn-road total:", on_road_total)
```

```python
rc_book = '1'
puc_certificate = '0'
service_history = '1'

rc_book_flag = bool(int(rc_book))
puc_certificate_flag = bool(int(puc_certificate))
service_history_flag = bool(int(service_history))

print("bool('0') vs bool(int('0')):", bool('0'), bool(int('0')))
print('rc_book: ' + rc_book + ' -> ' + str(rc_book_flag))
print('puc_certificate: ' + puc_certificate + ' -> ' + str(puc_certificate_flag))
print('service_history: ' + service_history + ' -> ' + str(service_history_flag))
```
