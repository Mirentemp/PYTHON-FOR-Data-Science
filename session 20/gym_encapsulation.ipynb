```python
class Member:
    def __init__(self, name, age, plan):
        self.name = name
        self.__age = age
        self.plan = plan
        self.visits = []
        if plan == 'Monthly':
            self.__fee_due = 1200
        elif plan == 'Quarterly':
            self.__fee_due = 3300
        elif plan == 'Annual':
            self.__fee_due = 11000
        else:
            self.__fee_due = 0

    def get_age(self):
        return self.__age

    def set_age(self, new_age):
        if not (14 <= new_age <= 80):
            raise ValueError('age must be between 14 and 80')
        self.__age = new_age

    def pay(self, amount):
        if amount <= 0 or amount > self.__fee_due:
            raise ValueError('invalid payment amount')
        self.__fee_due -= amount

    def balance_due(self):
        return self.__fee_due

q_member = Member('diya', 28, 'Quarterly')
q_member.pay(1000)
print('balance after 1000:', q_member.balance_due())
q_member.pay(2300)
print('balance after 2300:', q_member.balance_due())

try:
    print(q_member.__fee_due)
except AttributeError as e:
    print('attributeerror:', e)

print('mangled fee_due:', q_member._Member__fee_due)
```

```python
from datetime import datetime, timedelta

class GymDesk:
    def __calc_amount(self, member, months):
        if member.plan == 'Monthly':
            rate = 1200
        elif member.plan == 'Quarterly':
            rate = 1200
        elif member.plan == 'Annual':
            rate = 1200
        else:
            rate = 1200
        total = rate * months
        if months >= 6:
            total = total * 0.9
        return round(total, 2)

    def __valid_till_date(self, months):
        start_dt = datetime.strptime('2026-04-01', '%Y-%m-%d')
        end_dt = start_dt + timedelta(days=30 * months)
        return end_dt.strftime('%Y-%m-%d')

    def renew_membership(self, member, months):
        amt = self.__calc_amount(member, months)
        valid_till = self.__valid_till_date(months)
        return {
            'member': member.name,
            'plan': member.plan,
            'months': months,
            'amount': amt,
            'valid_till': valid_till
        }

desk = GymDesk()
m_renew = Member('aarav', 25, 'Monthly')
print(desk.renew_membership(m_renew, 1))
print(desk.renew_membership(m_renew, 3))
print(desk.renew_membership(m_renew, 6))
```
