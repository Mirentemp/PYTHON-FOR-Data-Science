```python
import csv

names = []
departments = []
amounts = []

with open('fest_registrations.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        names.append(row['name'])
        departments.append(row['department'])
        amounts.append(row['amount_inr'])

for idx, (n, d, a) in enumerate(zip(names, departments, amounts), start=1):
    print(f'{idx:02d} | {n} | {d} | rs {a}')
```

```python
import sys

def ticket_stream(n):
    for i in range(1, n + 1):
        yield 350 if i % 3 == 0 else 150

gen5 = ticket_stream(10)
print(next(gen5))
print(next(gen5))
print(next(gen5))
print(next(gen5))
print(next(gen5))

print(sys.getsizeof(list(ticket_stream(100000))), sys.getsizeof(ticket_stream(100000)))
```

```python
def running_total(stream):
    total = 0
    for amt in stream:
        total += amt
        yield total

for ticket_no, total in enumerate(running_total(ticket_stream(10000)), start=1):
    if total >= 50000:
        line = f'ticket {ticket_no} total {total}'
        print(line)
        with open('fest_milestone.txt', 'w', encoding='utf-8') as f:
            f.write(line + '\n')
        break
```
