```python
def delivery_charge(distance_km, base=20, per_km=6):
    if distance_km <= 2:
        return base
    return base + (distance_km - 2) * per_km

print(delivery_charge(1.5))
print(delivery_charge(5))
print(delivery_charge(5, per_km=8))
print(delivery_charge(distance_km=10, base=30))
```

```python
def weekly_summary(days, orders):
    total = sum(orders)
    avg = round(total / len(orders), 2)
    busiest = days[orders.index(max(orders))]
    return total, avg, busiest

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
orders = [30, 40, 35, 50, 45, 60, 25]

total_orders, avg_orders, busiest_day = weekly_summary(days, orders)
print(total_orders, avg_orders, busiest_day)

my_days = ['Fri', 'Sat', 'Sun']
my_orders = [42, 58, 35]
print(weekly_summary(my_days, my_orders))
```

```python
def tiffin_report(days, orders, price=80, discount_rate=0.0):
    total_orders, avg_orders, busiest_day = weekly_summary(days, orders)
    charge_5km = delivery_charge(5)
    revenue = total_orders * price * (1 - discount_rate)
    fees = total_orders * charge_5km
    print('revenue:', revenue)
    print('fees:', fees)

tiffin_report(days, orders)
tiffin_report(days, orders, discount_rate=0.1)
```
