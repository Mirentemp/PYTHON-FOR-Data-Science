```python
print('ghar tiffin service')

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
orders = [30, 40, 35, 50, 45, 60, 25]

total = 0
for day_num, day in enumerate(days, start=1):
    order_count = orders[day_num - 1]
    total = total + order_count
    print('Day ' + str(day_num) + ' ' + day + ': ' + str(order_count) + ' orders, running total ' + str(total))

avg = round(total / len(days), 2)
print(total)
print(avg)
```

```python
delivery_minutes = [28, 35, 0, 52, 41, 66, 33]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

scanned_total = 0
scanned_count = 0

for i in range(len(days)):
    mins = delivery_minutes[i]
    if mins == 0:
        continue
    if mins > 60:
        print('Late alert: ' + days[i] + ' (' + str(mins) + ' min)')
        break
    scanned_total = scanned_total + mins
    scanned_count = scanned_count + 1

print(scanned_total / scanned_count)
```
