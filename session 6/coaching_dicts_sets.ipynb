```python
class_profile = {
    101: {'name': 'Meet', 'subjects': {'Maths', 'Physics'}, 'fees_paid': 12000},
    102: {'name': 'Dhara', 'subjects': {'Maths', 'Biology'}, 'fees_paid': 24000},
    103: {'name': 'Naina', 'subjects': {'Physics'}, 'fees_paid': 18000},
    104: {'name': 'Ishita', 'subjects': {'Biology', 'English'}, 'fees_paid': 6000}
}

class_profile[101]['subjects'].add('Chemistry')
class_profile[104]['fees_paid'] = 12000
del class_profile[103]

total_fees = sum(student['fees_paid'] for student in class_profile.values())
all_subjects = set().union(*(student['subjects'] for student in class_profile.values()))

summary = {
    'total_fees': total_fees,
    'all_subjects': all_subjects
}
print(summary)
```
