```python
txns = [
    {'id': 'T001', 'merchant': ' sharma chai ', 'amount': 120, 'status': 'SUCCESS'},
    {'id': 'T002', 'merchant': 'CITY MEDICAL', 'amount': 2500, 'status': 'SUCCESS'},
    {'id': 'T003', 'merchant': ' patel bakery ', 'amount': 45, 'status': 'SUCCESS'},
    {'id': 'T004', 'merchant': 'ELECTRONICS HUB', 'amount': 18000, 'status': 'FAILED'},
    {'id': 'T005', 'merchant': ' fresh mart ', 'amount': 760, 'status': 'SUCCESS'},
    {'id': 'T006', 'merchant': 'BOOK WORLD', 'amount': 330, 'status': 'SUCCESS'},
    {'id': 'T007', 'merchant': ' metro furniture ', 'amount': 5200, 'status': 'SUCCESS'},
    {'id': 'T008', 'merchant': 'TEA POINT', 'amount': 90, 'status': 'FAILED'},
    {'id': 'T009', 'merchant': ' green pharmacy ', 'amount': 1499, 'status': 'SUCCESS'},
    {'id': 'T010', 'merchant': 'SNACK CORNER', 'amount': 60, 'status': 'SUCCESS'}
]

top3 = sorted(txns, key=lambda t: t['amount'], reverse=True)[:3]
print(list(map(lambda t: t['id'], top3)))
```

```python
clean_merchants = list(map(lambda t: t['merchant'].strip().title(), txns))
print(clean_merchants)

amounts_with_fee = list(map(lambda t: round(t['amount'] * 1.005, 2), txns))
print(amounts_with_fee)

high_success_txns = list(filter(lambda t: t['status'] == 'SUCCESS' and t['amount'] >= 500, txns))
print(high_success_txns)
print(list(map(lambda t: t['id'], high_success_txns)))
```

```python
from functools import reduce

success_txns = list(filter(lambda t: t['status'] == 'SUCCESS', txns))
success_total = reduce(lambda acc, t: acc + t['amount'], success_txns, 0)
success_max = reduce(lambda best, t: t['amount'] if t['amount'] > best else best, success_txns, 0)
sum_via_map = sum(map(lambda t: t['amount'], success_txns))

print(success_total)
print(success_max)
print(sum_via_map)
```
