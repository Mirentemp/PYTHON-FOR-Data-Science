```python
import csv

names = ['aarav', 'diya', 'kabir', 'ananya', 'rohan', 'meera', 'dev', 'kavya', 'arjun', 'tara', 'yash', 'riya']
depts = ['CSE', 'Mech', 'Civil', 'EC']

rows = []
for i in range(1, 13):
    reg_id = f'FEST{i:03d}'
    name = names[i - 1]
    dept = depts[(i - 1) % 4]
    ticket_type = 'Pass' if i % 3 == 0 else 'Day'
    amount_inr = 350 if ticket_type == 'Pass' else 150
    rows.append({
        'reg_id': reg_id,
        'name': name,
        'department': dept,
        'ticket_type': ticket_type,
        'amount_inr': amount_inr
    })

with open('fest_registrations.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['reg_id', 'name', 'department', 'ticket_type', 'amount_inr'])
    writer.writeheader()
    writer.writerows(rows)

print(len(rows))
```

```python
import csv

total_collection = 0
dept_counts = {}

with open('fest_registrations.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        total_collection += int(row['amount_inr'])
        d = row['department']
        dept_counts[d] = dept_counts.get(d, 0) + 1

with open('fest_summary.txt', 'w', encoding='utf-8') as f:
    f.write(f'total collection: {total_collection}\n')
    f.write(f'registrations per department: {dept_counts}\n')

with open('fest_summary.txt', 'a', encoding='utf-8') as f:
    f.write('prepared by miren\n')

with open('fest_summary.txt', 'r', encoding='utf-8') as f:
    print(f.read(), end='')

with open('fest_summary.txt', 'r', encoding='utf-8') as f:
    print(len(f.readlines()))
```

```python
import csv
import json

json_rows = []
with open('fest_registrations.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        json_rows.append({
            'reg_id': row['reg_id'],
            'name': row['name'],
            'department': row['department'],
            'ticket_type': row['ticket_type'],
            'amount_inr': int(row['amount_inr'])
        })

with open('fest_registrations.json', 'w', encoding='utf-8') as f:
    json.dump(json_rows, f, indent=2)

with open('fest_registrations.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)

pass_holders = [item['name'] for item in loaded if item['ticket_type'] == 'Pass']
print(pass_holders, len(pass_holders))
```
