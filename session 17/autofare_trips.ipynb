```python
from datetime import datetime
import statistics
import json

trips_raw = [
    ('2026-03-14 23:10', 6.2),
    ('2026-03-15 13:00', 1.2),
    ('2026-03-15 03:30', 4.0),
    ('2026-03-15 18:45', 8.5),
    ('2026-03-16 22:15', 3.5)
]

trip_records = []
fares = []

for dt_str, dist in trips_raw:
    dt_obj = datetime.strptime(dt_str, '%Y-%m-%d %H:%M')
    fare = autofare_utils.night_fare(dist, dt_obj.hour)
    fares.append(fare)
    print(dt_str, dist, fare)
    trip_records.append({
        'timestamp': dt_str,
        'distance_km': dist,
        'fare': fare
    })

print(statistics.mean(fares))
print(statistics.median(fares))

with open('autofare_trips.json', 'w', encoding='utf-8') as f:
    json.dump(trip_records, f, indent=2)
```
