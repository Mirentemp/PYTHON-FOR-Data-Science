```python
society_units = [[120, 135, 98], [210, [75, 80], 60], [150, [[40, 45], 30]]]

def total_units(block):
    total = 0
    for item in block:
        if isinstance(item, list):
            total += total_units(item)
        else:
            total += item
    return total

print(total_units(society_units))
print(total_units(society_units[0]))
print(total_units(society_units[1]))
print(total_units(society_units[2]))
```

```python
def nesting_depth(block, level=1):
    print('    ' * (level - 1) + str(block))
    max_depth = level
    for item in block:
        if isinstance(item, list):
            child_depth = nesting_depth(item, level + 1)
            if child_depth > max_depth:
                max_depth = child_depth
    return max_depth

print(nesting_depth([120, 135, 98]))
print(nesting_depth(society_units))
```

```python
RATE_PER_UNIT = 7
bills_generated = 0

def bill_for_wing(block):
    global bills_generated
    wing_bill = total_units(block) * RATE_PER_UNIT
    bills_generated += 1
    return wing_bill

b1 = bill_for_wing(society_units[0])
b2 = bill_for_wing(society_units[1])
b3 = bill_for_wing(society_units[2])

print(b1, b2, b3)
print(b1 + b2 + b3)
print(bills_generated)
```
