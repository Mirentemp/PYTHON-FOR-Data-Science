```python
import requests

response = requests.get('https://jsonplaceholder.typicode.com/todos')
print(response.status_code)

if response.status_code == 200:
    todos = response.json()
    print(type(todos), len(todos))
    for item in todos[:3]:
        print(item['title'])
else:
    print(response.text)
```

```python
print(f"{'userId':<8} {'total':<8} {'completed':<11} {'pct'}")
last_url = ''

for uid in range(1, 11):
    resp = requests.get('https://jsonplaceholder.typicode.com/todos', params={'userId': uid})
    last_url = resp.url
    items = resp.json()
    total = len(items)
    completed = sum(1 for t in items if t['completed'])
    pct = round((completed / total) * 100, 1)
    print(f"{uid:<8} {total:<8} {completed:<11} {pct}")

print(last_url)
```

```python
new_todos = [
    {'userId': 11, 'title': 'finish python api assignment', 'completed': 'false'},
    {'userId': 11, 'title': 'review pandas notebooks', 'completed': 'false'},
    {'userId': 11, 'title': 'push code to github', 'completed': 'false'}
]

for payload in new_todos:
    post_resp = requests.post('https://jsonplaceholder.typicode.com/todos', data=payload)
    print(post_resp.status_code, post_resp.json()['id'])

check_resp = requests.get('https://jsonplaceholder.typicode.com/todos/201')
print(check_resp.status_code)
```
